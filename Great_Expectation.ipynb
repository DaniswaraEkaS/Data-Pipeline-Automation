# Connect to Datasource

```python
# Give a name to a Datasource. This name must be unique between Datasources.
datasource_name = 'csv-data-job-engineer'
datasource = context.sources.add_pandas(datasource_name)

# Give a name to a data asset
asset_name = 'job-engineer-q1'
path_to_data = 'DataEngineerJob.csv'
asset = datasource.add_csv_asset(asset_name, filepath_or_buffer=path_to_data)

# Build batch request
batch_request = asset.build_batch_request()
```

```python
df = pd.read_csv(path_to_data)
df
```

# Create an Expectation Suite

```python
# Creat an expectation suite
expectation_suite_name = 'expectation-job-engineer-dataset'
context.add_or_update_expectation_suite(expectation_suite_name)

# Create a validator using above expectation suite
validator = context.get_validator(
    batch_request = batch_request,
    expectation_suite_name = expectation_suite_name
)

# Check the validator
validator.head()
```

# Expectations

```python
# Expectation 1 : Column `Job Title` can not contain missing values

validator.expect_column_values_to_not_be_null('Job Title')
```

```python
# Expectation 2 : Column `Salary Estimate` can not contain missing values

validator.expect_column_values_to_not_be_null('Salary Estimate')
```

```python
# Expectation 3 : Column `id` must be unique

validator.expect_column_values_to_be_unique('id')
```

```python
# Expectation 4 : Column `Rating` must be between `-1` and `5`

validator.expect_column_values_to_be_between(
    column='Rating', min_value=-1, max_value=5
)
```

```python
# Expectation 5 : Column `Company` Name must be exist so that job seekers know which 'Company' they will apply to.

validator.expect_column_to_exist(column='Company Name')
```

```python
# Expectation 6 : Column `Easy Apply` must contain one of the following 6 things :
# -1 = Not Easy To Apply
# True = Easy To Apply


validator.expect_column_values_to_be_in_set('Easy Apply', ["-1","True"])
```

```python
# Expectation 7 : Column `Rating` must in form of integer or float

validator.expect_column_values_to_be_in_type_list('Rating', ['integer', 'float'])
```

```python
# Expectation 8 : The maximum value of column `Founded` must be `2025`

validator.expect_column_max_to_be_between('Founded', 0, 2025)
```

```python
# Save into Expectation Suite

validator.save_expectation_suite(discard_failed_expectations=False)
```

By default, Great Expectation will only store successful Expectations. Set param `discard_failed_expectation=False` so that all Expectations both successful and failed to be stored.

## Checkpoint

```python
# Create a checkpoint

checkpoint_1 = context.add_or_update_checkpoint(
    name = 'checkpoint_1',
    validator = validator,
)
```

```python
# Run a checkpoint

checkpoint_result = checkpoint_1.run()
```
